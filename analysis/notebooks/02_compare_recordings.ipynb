# Compare recordings

Tick two or more recordings in the list that opens (Ctrl or Shift to select several). Use it to compare motors on the same propeller, or propellers on the same motor.

```python
import sys, pathlib
# Makes the package importable when the notebook is run from the repository.
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import thrustbench_analysis as tba
```

```python
recs = tba.pick_recordings()
for r in recs:
    print(r.label, '-', r.path.name)
```

## Throttle curves

```python
tba.plot_comparison(recs);
```

```python
import pandas as pd
pd.concat({r.label: tba.throttle_curve(r).set_index('throttle_pct') for r in recs}, axis=1).round(3)
```

## Spectra

```python
tba.plot_spectrum(recs, 'thrust');
```
