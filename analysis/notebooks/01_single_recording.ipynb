# One recording

Run the cells in order. Nothing needs editing: the recording is chosen in a dialog, and names, units and calibration are read from the recording itself.

If the dialog does not appear, look behind the browser window.

```python
import sys, pathlib
# Makes the package importable when the notebook is run from the repository.
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import thrustbench_analysis as tba
```

```python
rec = tba.pick_recording()
print(rec.label)
print(rec.path)
```

## Overview
Every channel against time. Red lines mark events stored with the recording (emergency stop, link loss, profile start...).

```python
tba.plot_overview(rec);
rec.events
```

## Statistics

```python
tba.describe(rec)
```

## Throttle curve
Needs a recording made with a staircase profile and the ESC driven by the application.

```python
curve = tba.throttle_curve(rec)
tba.plot_curve(rec);
curve
```

## Spectrum
The load cells are sampled at 80 Hz, so this spectrum ends at 40 Hz.

```python
tba.plot_spectrum(rec, 'thrust');
```
